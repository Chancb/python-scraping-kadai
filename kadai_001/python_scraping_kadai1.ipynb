```python
import requests
from bs4 import BeautifulSoup

url = 'https://news.yahoo.co.jp/articles/52ef13ead5db01ec3f4ebbbac8e6473a8f73f776'
response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')

# <p> をすべて取得（CSSセレクタ）
article_elements = soup.select('#uamods > div p')

for article_element in article_elements:
    # 直下テキストだけ取得（aタグの中身は除外）
    article_text = ''.join(article_element.find_all(string=True, recursive=False)).strip()
    print(article_text)
```
