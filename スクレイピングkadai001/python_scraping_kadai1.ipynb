```python
import requests
from bs4 import BeautifulSoup
url='https://news.yahoo.co.jp/articles/52ef13ead5db01ec3f4ebbbac8e6473a8f73f776'
response=requests.get(url)
# HTMLを解析する
soup = BeautifulSoup(response.text, 'html.parser')
# articleの要素を取得する
# CSSセレクタを使用して、articleを取得する
article_element = soup.select_one('#uamods >div ')
# articleのテキストを取得する
article_text = article_element.text
print(article_text)
```
